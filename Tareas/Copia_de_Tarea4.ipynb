## <font color='8EC044'> **Ejercicio 1** </font>

Diseñe una función que dados el argumento `material` retorne un mapa de colombia visualizando todos los transformadores que realizan procesamiento de dicho material. La función deberá tener un valor por defecto todos, que muestre todos los transformadores sin impoertar le material que procesan. Los posible valores para el argunmento material son `['plástico', 'cartón', 'vidrio', 'metales', 'multimaterial', 'varios', 'papel']`


```python
# Escriba aquí su respuesta
import geopandas as gpd
import pandas as pd
import plotly.express as px

def map_transformadores_por_material(df, material='todos'):

    # Verificar que el material sea válido
    materiales_validos = ['todos', 'plástico', 'cartón', 'vidrio', 'metales', 'multimaterial', 'varios', 'papel']
    if material not in materiales_validos:
        raise ValueError(f"Material inválido. Opciones disponibles: {materiales_validos}")

    # Filtrar el dataframe según el material seleccionado
    if material != 'todos':
        df_filtrado = df[df[material] == 1]
    else:
        df_filtrado = df.copy()

    # Verificar que haya datos para el material seleccionado
    if df_filtrado.empty:
        print(f"No se encontraron transformadores para el material '{material}'.")
        return

    # Crear el mapa interactivo usando Plotly Express
    fig = px.scatter_mapbox(
        df_filtrado,
        lat="latitud",
        lon="longitud",
        hover_name="razon social",
        hover_data={"departamento": True, "latitud": False, "longitud": False},
        color_discrete_sequence=["blue"],
        zoom=5,
        height=600
    )

    fig.update_layout(
        mapbox_style="carto-positron",
        mapbox_center={"lat": 4.5709, "lon": -74.2973},  # Coordenadas aproximadas del centro de Colombia
        title=f"Transformadores de {material.title() if material != 'todos' else 'Todos los materiales'} en Colombia"
    )

    fig.show()
```

Pruebe que la función funciona corriendola para un ejemplo en el DataFrame `df`

```python
# Escriba aquí su respuesta
map_transformadores_por_material(df, material='plástico')
```

## <font color='8EC044'> **Ejercicio 2** </font>

Diseñe una función que dados el argumento `material` retorne un gráfico de barras que retorne la capacidad de procesamiento de dicho material en los distintos departamentos. La función deberá tener un valor por defecto todos, que muestre todos los transformadores sin impoertar le material que procesan


```python
import pandas as pd
import matplotlib.pyplot as plt


def capacidad_por_departamento(df, material='todos'):

    if material != 'todos':
        df_filtered = df[df['material'] == material]
    else:
        df_filtered = df

    capacidad_por_departamento = df_filtered.groupby('departamento')['capacidad'].sum()

    plt.figure(figsize=(12, 8))
    capacidad_por_departamento.plot(kind='bar', color='skyblue')
    plt.title(f"Capacidad de procesamiento por departamento para {material.capitalize() if material != 'todos' else 'todos los materiales'}")
    plt.xlabel("Departamento")
    plt.ylabel("Capacidad de procesamiento")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


```

```python

capacidad_por_departamento(df)  # Para todos los materiales
```

```python
import pandas as pd
import plotly.express as px

def grafico_capacidad_por_departamento(df, material='todos'):

    # Verificar que el material sea válido
    materiales_validos = ['todos', 'plástico', 'cartón', 'vidrio', 'metales', 'multimaterial', 'varios', 'papel']
    if material not in materiales_validos:
        raise ValueError(f"Material inválido. Opciones disponibles: {materiales_validos}")

    # Filtrar el dataframe según el material seleccionado
    if material != 'todos':
        df_filtrado = df[df[material] == 1]
    else:
        df_filtrado = df.copy()

    # Verificar que haya datos para el material seleccionado
    if df_filtrado.empty:
        print(f"No se encontraron datos de capacidad para el material '{material}'.")
        return

    # Agrupar por departamento y calcular la capacidad total
    capacidad_por_departamento = df_filtrado.groupby('departamento')['capacidad'].sum().reset_index()

    # Crear el gráfico de barras usando Plotly Express
    fig = px.bar(
        capacidad_por_departamento,
        x='departamento',
        y='capacidad',
        title=f"Capacidad de procesamiento de {material.title() if material != 'todos' else 'Todos los materiales'} por Departamento",
        labels={'capacidad': 'Capacidad de Procesamiento', 'departamento': 'Departamento'},
        color_discrete_sequence=['blue']
    )

    fig.update_layout(xaxis={'categoryorder': 'total descending'})
    fig.show()
```

```python

grafico_capacidad_por_departamento(df, material='plástico')
```
